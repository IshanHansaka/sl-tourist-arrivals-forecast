# Data Preprocessing and Feature Engineering

This notebook uses a leak-free, shared feature implementation from `src/features.py`.

```python
import sys
sys.path.insert(0, '..')

from src.data_processing import load_raw_data, clean_data, save_cleaning_artifacts
from src.features import build_features, chronological_split, fit_country_tiers, save_feature_metadata

raw = load_raw_data('../data/raw/sl_tourist_arrivals_dataset.csv')
cleaned, audit = clean_data(raw)
save_cleaning_artifacts(raw, cleaned, audit, '../data/processed')

splits = chronological_split(cleaned)
train = splits['train']
print({name: (len(part), str(part['Date'].min().date()), str(part['Date'].max().date())) for name, part in splits.items()})

country_tiers = fit_country_tiers(train)
featured = build_features(cleaned, country_tiers)
save_feature_metadata('../artifacts/feature_metadata.json', country_tiers)
print(featured.filter(like='Lag_').shape, len(country_tiers))
```
